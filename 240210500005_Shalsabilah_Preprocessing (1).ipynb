```python
import re
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# Inisialisasi stemmer dan stopword remover Sastrawi
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()

stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()
stopwords_list = set(stopword_factory.get_stop_words())

print("Library berhasil dimuat.")
```

```python
documents = [
    "Fakultas Teknik UNM mengumumkan bahwa pendaftaran Kuliah Kerja Nyata (KKN) 2026 telah dibuka mulai hari ini! Mahasiswa diharapkan segera mendaftarkan diri melalui portal akademik.",
    "Perkembangan teknologi kecerdasan buatan (AI) di Indonesia semakin pesat pada tahun 2026. Banyak perusahaan mulai mengimplementasikan sistem berbasis AI untuk meningkatkan efisiensi kerja mereka.",
    "Laboratorium Komputer Fakultas Teknik akan melakukan pemeliharaan rutin pada tanggal 25 September 2026. Selama proses ini, seluruh layanan laboratorium tidak dapat diakses oleh mahasiswa.",
    "Sistem temu kembali informasi merupakan salah satu cabang ilmu komputer yang mempelajari bagaimana cara menemukan dokumen relevan dari kumpulan data yang besar secara cepat dan akurat.",
    "Program studi Teknik Komputer membuka kesempatan magang bagi mahasiswa semester akhir di berbagai perusahaan teknologi terkemuka, baik di dalam maupun luar negeri.",
]

for i, doc in enumerate(documents, 1):
    print(f"Dokumen {i}: {doc}\n")
```

```python
def preprocess_text(text):
    """
    Melakukan pipeline preprocessing teks Bahasa Indonesia:
    1. Case folding
    2. Cleaning (hapus angka, tanda baca, karakter khusus)
    3. Tokenisasi
    4. Stopwords removal (Sastrawi)
    5. Stemming (Sastrawi)

    Return:
        tokens_before (list) : token hasil tokenisasi awal (sebelum stopword & stemming)
        tokens_after  (list) : token akhir setelah full pipeline
    """
    # 1. Case folding
    text = text.lower()

    # 2. Cleaning: hapus angka, tanda baca, dan karakter khusus
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # 3. Tokenisasi
    tokens_before = text.split()

    # 4. Stopwords removal
    tokens_no_stopwords = [t for t in tokens_before if t not in stopwords_list]

    # 5. Stemming (Sastrawi bekerja pada string, jadi digabung lalu di-stem lalu di-split lagi)
    stemmed_text = stemmer.stem(' '.join(tokens_no_stopwords))
    tokens_after = stemmed_text.split()

    return tokens_before, tokens_after

# Uji cepat fungsi
test_before, test_after = preprocess_text(documents[0])
print("Sebelum:", test_before)
print("Sesudah:", test_after)
```

```python
results = []
for i, doc in enumerate(documents, 1):
    tokens_before, tokens_after = preprocess_text(doc)
    results.append({
        "no": i,
        "dokumen_asli": doc,
        "tokens_before": tokens_before,
        "tokens_after": tokens_after,
        "jumlah_token_sebelum": len(tokens_before),
        "jumlah_token_sesudah": len(tokens_after),
    })

print("Preprocessing selesai untuk", len(results), "dokumen.")
```

```python
# Tampilkan 2 dokumen pertama secara lengkap
for r in results[:2]:
    print(f"===== Dokumen {r['no']} =====")
    print("Teks asli       :", r["dokumen_asli"])
    print("Token sebelum   :", r["tokens_before"])
    print("Token sesudah   :", r["tokens_after"])
    print()
```

```python
# Ringkasan seluruh dokumen (tabel)
for r in results:
    print(f"Dokumen {r['no']}")
    print("  Sebelum :", " ".join(r["tokens_before"]))
    print("  Sesudah :", " ".join(r["tokens_after"]))
    print()
```

```python
summary_df = pd.DataFrame([
    {
        "Dokumen": r["no"],
        "Jumlah Token Sebelum": r["jumlah_token_sebelum"],
        "Jumlah Token Sesudah": r["jumlah_token_sesudah"],
        "Pengurangan Token": r["jumlah_token_sebelum"] - r["jumlah_token_sesudah"],
        "Persentase Pengurangan (%)": round(
            (r["jumlah_token_sebelum"] - r["jumlah_token_sesudah"]) / r["jumlah_token_sebelum"] * 100, 2
        ),
    }
    for r in results
])

total_before = summary_df["Jumlah Token Sebelum"].sum()
total_after = summary_df["Jumlah Token Sesudah"].sum()
total_reduction_pct = round((total_before - total_after) / total_before * 100, 2)

summary_df
```

```python
print(f"Total token sebelum preprocessing : {total_before}")
print(f"Total token sesudah preprocessing : {total_after}")
print(f"Total persentase pengurangan token : {total_reduction_pct}%")
```

Analisis Singkat :
Preprocessing (case folding, cleaning, stopwords removal, dan stemming) terbukti mengurangi jumlah token secara signifikan dengan menghilangkan kata-kata yang tidak bermakna secara informasional (stopwords) dan menyeragamkan variasi bentuk kata berimbuhan menjadi kata dasar melalui stemming. Bagi sistem temu kembali informasi (IR), hal ini meningkatkan kualitas data karena representasi dokumen menjadi lebih ringkas, konsisten, dan fokus pada term-term yang benar-benar merepresentasikan isi/topik dokumen, sehingga proses pencocokan (matching) antara query dan dokumen menjadi lebih akurat, mengurangi noise, dan mempercepat komputasi pada tahap pembobotan maupun perhitungan kemiripan (similarity) selanjutnya.
